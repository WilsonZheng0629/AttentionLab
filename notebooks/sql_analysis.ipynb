```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("/Users/wilson/AttentionLab/data/attentionlab.db")

```

```python
# Top 10 videos by views    

top_videos = pd.read_sql("""
SELECT
    caption,
    platform,
    topic,
    format,
    views
FROM videos
ORDER BY views DESC
LIMIT 10;
""", conn)

top_videos
```

```python
# 10 videos by engagement rate (likes + comments / views)  
best_platform = pd.read_sql("""
SELECT
    platform,
    COUNT(*) AS video_count,
    ROUND(AVG(views), 2) AS avg_views,
    SUM(views) AS total_views
FROM videos
GROUP BY platform
ORDER BY avg_views DESC;
""", conn)

best_platform
```

```python
# Best platform by average views
best_topic = pd.read_sql("""
SELECT
    topic,
    COUNT(*) AS video_count,
    ROUND(AVG(views), 2) AS avg_views,
    SUM(views) AS total_views
FROM videos
GROUP BY topic
ORDER BY avg_views DESC;
""", conn)

best_topic
```

```python
# Best format for engagement rate
best_format = pd.read_sql("""
SELECT
    format,
    COUNT(*) AS video_count,
    ROUND(AVG(engagement_rate), 4) AS avg_engagement_rate,
    ROUND(AVG(views), 2) AS avg_views
FROM videos
GROUP BY format
ORDER BY avg_engagement_rate DESC;
""", conn)

best_format
```

```python
# Best hook type by engagment rate   
best_hook = pd.read_sql("""
SELECT
    hook_type,
    COUNT(*) AS video_count,
    ROUND(AVG(views), 2) AS avg_views,
    ROUND(AVG(engagement_rate), 4) AS avg_engagement_rate
FROM videos
GROUP BY hook_type
ORDER BY avg_views DESC;
""", conn)

best_hook
```

```python
# Best video legnth bucket by engagement rate
best_length = pd.read_sql("""
SELECT
    length_bucket,
    COUNT(*) AS video_count,
    ROUND(AVG(views), 2) AS avg_views,
    ROUND(AVG(engagement_rate), 4) AS avg_engagement_rate
FROM videos
GROUP BY length_bucket
ORDER BY avg_views DESC;
""", conn)

best_length
```

```python
# Best day of the week by engagement rate
best_day = pd.read_sql("""
SELECT
    day_of_week,
    COUNT(*) AS video_count,
    ROUND(AVG(views), 2) AS avg_views,
    ROUND(AVG(engagement_rate), 4) AS avg_engagement_rate
FROM videos
GROUP BY day_of_week
ORDER BY avg_views DESC;
""", conn)

best_day
```

```python
# Weekday vs Weekend by engagement rate
weekend_analysis = pd.read_sql("""
SELECT
    is_weekend,
    COUNT(*) AS video_count,
    ROUND(AVG(views), 2) AS avg_views,
    ROUND(AVG(engagement_rate), 4) AS avg_engagement_rate
FROM videos
GROUP BY is_weekend;
""", conn)

weekend_analysis
```

```python
# Production ROI by views
production_roi = pd.read_sql("""
SELECT
    caption,
    platform,
    topic,
    views,
    editing_time_minutes,
    ROUND(views_per_minute, 2) AS views_per_minute
FROM videos
ORDER BY views_per_minute DESC
LIMIT 10;
""", conn)

production_roi
```

```python
# Best topic by follower conversion 
follower_conversion = pd.read_sql("""
SELECT
    topic,
    COUNT(*) AS video_count,
    ROUND(AVG(follower_conversion_rate), 4) AS avg_follower_conversion_rate,
    SUM(followers_gained) AS total_followers_gained
FROM videos
GROUP BY topic
ORDER BY avg_follower_conversion_rate DESC;
""", conn)

follower_conversion
```

```python
# Low-effort high-return videos
low_effort_high_return = pd.read_sql("""
SELECT
    caption,
    platform,
    topic,
    views,
    editing_time_minutes,
    ROUND(views_per_minute, 2) AS views_per_minute
FROM videos
WHERE editing_time_minutes <= 35
ORDER BY views_per_minute DESC
LIMIT 10;
""", conn)

low_effort_high_return
```
